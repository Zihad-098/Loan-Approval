```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

from dataclasses import dataclass
from typing import List, Dict, Tuple, Optional

from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler, PowerTransformer, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.inspection import permutation_importance
from sklearn.metrics import roc_auc_score, accuracy_score


# ==========================
# Utility helpers
# ==========================
def _minmax01(x: np.ndarray) -> np.ndarray:
    x = np.asarray(x, dtype=float)
    mn, mx = np.min(x), np.max(x)
    return (x - mn) / (mx - mn + 1e-12)

def _sigmoid(x: np.ndarray) -> np.ndarray:
    return 1.0 / (1.0 + np.exp(-x))

def _mad(x: np.ndarray) -> float:
    x = np.asarray(x, dtype=float)
    med = np.median(x)
    return float(np.median(np.abs(x - med)) + 1e-12)

def _is_binary(y: np.ndarray) -> bool:
    uniq = np.unique(y)
    return len(uniq) == 2

def _safe_auc(y_true, y_score) -> float:
    try:
        return roc_auc_score(y_true, y_score)
    except Exception:
        return 0.5

def _wrap_by_underscore(label: str, max_lines: int = 3) -> str:
    """
    Nicely wrap feature names like:
      'person_home_ownership' -> 'person\nhome\nownership'
    If too many parts, keep last parts merged so it doesn't become too tall.
    """
    parts = str(label).split("_")
    if len(parts) <= 1:
        return str(label)

    if len(parts) <= max_lines:
        return "\n".join(parts)

    # If too many parts, merge the tail
    head = parts[: max_lines - 1]
    tail = "_".join(parts[max_lines - 1 :])
    return "\n".join(head + [tail])


# ==========================
# Config
# ==========================
@dataclass
class HybridFSConfig:
    # Embedded stability selector
    stability_bootstraps: int = 30
    stability_subsample: float = 0.75
    elasticnet_C: float = 1.0
    elasticnet_l1_ratio: float = 0.7
    coef_eps: float = 1e-6

    # AutoML-style selector (lightweight)
    automl_folds: int = 3
    perm_repeats: int = 5
    beta_softmax: float = 6.0

    # GA selector
    ga_folds: int = 3
    ga_pop_size: int = 24
    ga_generations: int = 12
    ga_tournament_k: int = 3
    ga_elite_frac: float = 0.25
    ga_mutation_rate: Optional[float] = None  # default -> 1/d
    ga_crossover_rate: float = 0.9
    ga_lambda_sparsity: float = 0.15
    ga_gamma_weight: float = 6.0

    # Fusion (novel)
    fusion_eta_var: float = 0.25
    fusion_kappa: float = 0.35
    tau_min: float = 0.50
    tau_max: float = 0.92

    # Final selection caps
    min_features: int = 8
    max_features: int = 20

    # Plotting
    fig_dpi: int = 220
    font_size: int = 11
    heatmap_cmap: str = "viridis"


# ==========================
# Main Hybrid Selector
# ==========================
class ICMLHybridFeatureSelector:
    def __init__(self, cfg: HybridFSConfig = HybridFSConfig(), random_state: int = 42):
        self.cfg = cfg
        self.random_state = random_state

        self.feature_names_: List[str] = []
        self.emb_score_: Optional[np.ndarray] = None
        self.aml_score_: Optional[np.ndarray] = None
        self.ga_score_: Optional[np.ndarray] = None
        self.fused_score_: Optional[np.ndarray] = None

        self.emb_selected_: List[int] = []
        self.aml_selected_: List[int] = []
        self.ga_selected_: List[int] = []
        self.selected_: List[int] = []

        self.model_weights_: Dict[str, float] = {}

    # -------- Embedded stability selector --------
    def _embedded_stability(self, X: np.ndarray, y: np.ndarray) -> np.ndarray:
        rng = np.random.default_rng(self.random_state)

        pt = PowerTransformer(method="yeo-johnson", standardize=True)
        Xt = pt.fit_transform(X)

        n, d = Xt.shape
        B = self.cfg.stability_bootstraps
        subsz = int(max(10, round(self.cfg.stability_subsample * n)))

        select_counts = np.zeros(d, dtype=float)
        coef_mags = np.zeros(d, dtype=float)

        for b in range(B):
            idx = rng.choice(n, size=subsz, replace=False)
            Xb, yb = Xt[idx], y[idx]

            # no n_jobs -> avoids sklearn/joblib pos_class error
            clf = LogisticRegression(
                solver="saga",
                penalty="elasticnet",
                l1_ratio=self.cfg.elasticnet_l1_ratio,
                C=self.cfg.elasticnet_C,
                max_iter=5000,
                random_state=self.random_state + b,
            )
            clf.fit(Xb, yb)

            w = np.ravel(clf.coef_)
            coef_mags += np.abs(w)
            select_counts += (np.abs(w) > self.cfg.coef_eps).astype(float)

        pi = select_counts / (B + 1e-12)
        coef_mags = coef_mags / (B + 1e-12)

        score = 0.65 * _minmax01(pi) + 0.35 * _minmax01(coef_mags)
        self.emb_score_ = score
        return score

    # -------- AutoML selector --------
    def _automl_weighted_pfi(self, X: np.ndarray, y: np.ndarray, scoring: str) -> np.ndarray:
        d = X.shape[1]
        kf = StratifiedKFold(n_splits=self.cfg.automl_folds, shuffle=True, random_state=self.random_state)

        candidates = [
            ("LogReg_L2", LogisticRegression(solver="lbfgs", penalty="l2", max_iter=4000)),  # no n_jobs
            ("RF", RandomForestClassifier(n_estimators=400, random_state=self.random_state, n_jobs=-1)),
            ("ExtraTrees", ExtraTreesClassifier(n_estimators=600, random_state=self.random_state, n_jobs=-1)),
            ("HGB", HistGradientBoostingClassifier(random_state=self.random_state)),
        ]

        model_scores = []
        model_importances = []

        for name, model in candidates:
            fold_scores = []
            fold_imps = []

            for tr, va in kf.split(X, y):
                Xtr, Xva = X[tr], X[va]
                ytr, yva = y[tr], y[va]

                model.fit(Xtr, ytr)

                if scoring == "roc_auc":
                    prob = model.predict_proba(Xva)[:, 1] if hasattr(model, "predict_proba") else model.decision_function(Xva)
                    sc = _safe_auc(yva, prob)
                else:
                    pred = model.predict(Xva)
                    sc = accuracy_score(yva, pred)

                fold_scores.append(sc)

                pi = permutation_importance(
                    model, Xva, yva,
                    n_repeats=self.cfg.perm_repeats,
                    random_state=self.random_state,
                    scoring=scoring
                ).importances_mean
                fold_imps.append(np.maximum(0.0, pi))

            mean_score = float(np.mean(fold_scores))
            mean_imp = np.mean(np.vstack(fold_imps), axis=0)

            model_scores.append(mean_score)
            model_importances.append(mean_imp)

        model_scores = np.array(model_scores, dtype=float)

        beta = self.cfg.beta_softmax
        w = np.exp(beta * (model_scores - np.max(model_scores)))
        w = w / (np.sum(w) + 1e-12)

        agg = np.zeros(d, dtype=float)
        for wm, imp in zip(w, model_importances):
            agg += float(wm) * _minmax01(imp)

        self.aml_score_ = _minmax01(agg)

        self.model_weights_["AutoML_LogReg_L2"] = float(w[0])
        self.model_weights_["AutoML_RF"] = float(w[1])
        self.model_weights_["AutoML_ExtraTrees"] = float(w[2])
        self.model_weights_["AutoML_HGB"] = float(w[3])

        return self.aml_score_

    # -------- GA selector --------
    def _ga_feature_selection(self, X: np.ndarray, y: np.ndarray, scoring: str) -> np.ndarray:
        rng = np.random.default_rng(self.random_state)
        n, d = X.shape

        kf = StratifiedKFold(n_splits=self.cfg.ga_folds, shuffle=True, random_state=self.random_state)
        base = LogisticRegression(solver="lbfgs", max_iter=4000)  # no n_jobs

        lam = self.cfg.ga_lambda_sparsity
        mut_rate = self.cfg.ga_mutation_rate if self.cfg.ga_mutation_rate is not None else (1.0 / d)

        cache: Dict[bytes, float] = {}

        def cv_score(mask: np.ndarray) -> float:
            if mask.sum() == 0:
                return 0.0

            key = mask.tobytes()
            if key in cache:
                return cache[key]

            Xs = X[:, mask.astype(bool)]
            fold_scores = []
            for tr, va in kf.split(Xs, y):
                Xtr, Xva = Xs[tr], Xs[va]
                ytr, yva = y[tr], y[va]
                base.fit(Xtr, ytr)

                if scoring == "roc_auc":
                    prob = base.predict_proba(Xva)[:, 1]
                    fold_scores.append(_safe_auc(yva, prob))
                else:
                    pred = base.predict(Xva)
                    fold_scores.append(accuracy_score(yva, pred))

            sc = float(np.mean(fold_scores))
            cache[key] = sc
            return sc

        def fitness(mask: np.ndarray) -> float:
            sc = cv_score(mask)
            spars = float(mask.sum()) / float(d)
            return sc - lam * spars

        target = min(max(self.cfg.min_features, int(np.sqrt(n))), self.cfg.max_features)
        p_on = target / max(d, 1)

        pop = (rng.random((self.cfg.ga_pop_size, d)) < p_on).astype(np.int8)
        for i in range(pop.shape[0]):
            if pop[i].sum() == 0:
                pop[i, rng.integers(0, d)] = 1

        def tournament_select(pop, fitvals):
            idxs = rng.integers(0, len(pop), size=self.cfg.ga_tournament_k)
            best = idxs[np.argmax(fitvals[idxs])]
            return pop[best].copy()

        def crossover(a, b):
            if rng.random() > self.cfg.ga_crossover_rate:
                return a.copy(), b.copy()
            mask = rng.random(d) < 0.5
            c1 = np.where(mask, a, b).astype(np.int8)
            c2 = np.where(mask, b, a).astype(np.int8)
            if c1.sum() == 0: c1[rng.integers(0, d)] = 1
            if c2.sum() == 0: c2[rng.integers(0, d)] = 1
            return c1, c2

        def mutate(x):
            flips = rng.random(d) < mut_rate
            x = x.copy()
            x[flips] = 1 - x[flips]
            if x.sum() == 0:
                x[rng.integers(0, d)] = 1
            return x

        for _ in range(self.cfg.ga_generations):
            fitvals = np.array([fitness(ind) for ind in pop], dtype=float)

            elite_n = max(1, int(self.cfg.ga_elite_frac * len(pop)))
            elite_idx = np.argsort(fitvals)[-elite_n:]
            elites = pop[elite_idx].copy()

            new_pop = [elites[i] for i in range(elite_n)]
            while len(new_pop) < len(pop):
                p1 = tournament_select(pop, fitvals)
                p2 = tournament_select(pop, fitvals)
                c1, c2 = crossover(p1, p2)
                c1 = mutate(c1)
                if len(new_pop) < len(pop):
                    new_pop.append(c1)
                if len(new_pop) < len(pop):
                    c2 = mutate(c2)
                    new_pop.append(c2)

            pop = np.array(new_pop[:len(pop)], dtype=np.int8)

        fitvals = np.array([fitness(ind) for ind in pop], dtype=float)
        elite_n = max(2, int(self.cfg.ga_elite_frac * len(pop)))
        elite_idx = np.argsort(fitvals)[-elite_n:]
        elites = pop[elite_idx].astype(float)
        elite_fit = fitvals[elite_idx]

        gamma = self.cfg.ga_gamma_weight
        w = np.exp(gamma * (elite_fit - np.max(elite_fit)))
        w = w / (np.sum(w) + 1e-12)

        p_ga = (w.reshape(-1, 1) * elites).sum(axis=0)
        self.ga_score_ = _minmax01(p_ga)
        return self.ga_score_

    # -------- Fusion --------
    def _fuse(self, emb: np.ndarray, aml: np.ndarray, ga: np.ndarray, n: int, d: int) -> Tuple[np.ndarray, float]:
        e = _minmax01(emb)
        a = _minmax01(aml)
        g = _minmax01(ga)

        global_strength = np.array([e.mean(), a.mean(), g.mean()], dtype=float)
        w = np.exp(5.0 * (global_strength - global_strength.max()))
        w = w / (w.sum() + 1e-12)
        we, wa, wg = map(float, w)

        var = np.var(np.vstack([e, a, g]), axis=0)

        fused = _sigmoid(we * e + wa * a + wg * g - self.cfg.fusion_eta_var * var)
        fused = _minmax01(fused)

        tau = np.quantile(fused, 0.75) + self.cfg.fusion_kappa * _mad(fused) * np.sqrt(np.log(max(d, 2)) / max(n, 5))
        tau = float(np.clip(tau, self.cfg.tau_min, self.cfg.tau_max))

        return fused, tau

    def fit(self, X: np.ndarray, y: np.ndarray, feature_names: List[str]):
        self.feature_names_ = list(feature_names)
        scoring = "roc_auc" if _is_binary(y) else "accuracy"

        emb = self._embedded_stability(X, y)
        aml = self._automl_weighted_pfi(X, y, scoring=scoring)
        ga = self._ga_feature_selection(X, y, scoring=scoring)

        fused, tau = self._fuse(emb, aml, ga, n=X.shape[0], d=X.shape[1])
        self.fused_score_ = fused

        idx = np.where(fused >= tau)[0].tolist()

        if len(idx) < self.cfg.min_features:
            idx = np.argsort(fused)[-self.cfg.min_features:].tolist()
        if len(idx) > self.cfg.max_features:
            idx = np.argsort(fused)[-self.cfg.max_features:].tolist()

        self.selected_ = sorted(idx)

        def topk(score, k):
            k = min(k, len(score))
            return np.argsort(score)[-k:].tolist()

        k_ref = min(self.cfg.max_features, max(self.cfg.min_features, 12))
        self.emb_selected_ = topk(emb, k_ref)
        self.aml_selected_ = topk(aml, k_ref)
        self.ga_selected_ = topk(ga, k_ref)

        self.model_weights_["Fusion_tau"] = float(tau)
        return self

    def transform(self, X: np.ndarray) -> np.ndarray:
        return X[:, self.selected_]

    def fit_transform(self, X: np.ndarray, y: np.ndarray, feature_names: List[str]) -> np.ndarray:
        return self.fit(X, y, feature_names).transform(X)

    # -------- plotting (FIXED LABEL OVERLAP) --------
    def plot_heatmap(self, save_path="hybrid_feature_heatmap.png", show_plot=True):
        if self.fused_score_ is None:
            print("Call fit() first.")
            return None

        plt.rcParams["axes.grid"] = False

        sel = self.selected_
        cols = [self.feature_names_[i] for i in sel]
        cols_wrapped = [_wrap_by_underscore(c, max_lines=3) for c in cols]

        mat = np.vstack([
            _minmax01(self.emb_score_)[sel],
            _minmax01(self.aml_score_)[sel],
            _minmax01(self.ga_score_)[sel],
            _minmax01(self.fused_score_)[sel],
        ])
        df = pd.DataFrame(mat, index=["Embedded", "AutoML-PFI", "GA", "Fused(Novel)"], columns=cols_wrapped)

        # Wider figure per feature + extra bottom margin
        fig_w = max(12, 1.35 * len(cols_wrapped))
        fig_h = 5.5
        plt.figure(figsize=(fig_w, fig_h), dpi=self.cfg.fig_dpi)

        cmap = sns.color_palette(self.cfg.heatmap_cmap, as_cmap=True)
        ax = sns.heatmap(
            df,
            annot=True, fmt=".2f",
            cmap=cmap,
            linewidths=0.8, linecolor="white",
            cbar_kws={"shrink": 0.85},
            annot_kws={"size": 11, "weight": "bold"},
        )

        ax.grid(False)

        plt.title("Hybrid Feature Scores (Embedded + AutoML + GA → Novel Fusion)", fontsize=15, fontweight="bold", pad=14)
        plt.xlabel("Selected Features", fontsize=12, fontweight="bold")
        plt.ylabel("Method", fontsize=12, fontweight="bold")

        # KEY FIX: rotate + right align so wrapped labels never collide
        plt.xticks(rotation=35, ha="right", fontsize=10)
        plt.yticks(rotation=0, fontsize=11)

        # more bottom space so wrapped/rotated labels don't overwrite each other
        plt.subplots_adjust(bottom=0.35)

        plt.tight_layout()
        plt.savefig(save_path, dpi=300, bbox_inches="tight")
        if show_plot:
            plt.show()
        else:
            plt.close()
        print(f"Saved heatmap: {save_path}")
        return df

    def plot_fused_bar(self, save_path="fused_feature_scores.png", show_plot=True):
        if self.fused_score_ is None:
            print("Call fit() first.")
            return

        sel = self.selected_
        names = [self.feature_names_[i] for i in sel]
        scores = self.fused_score_[sel]

        order = np.argsort(scores)
        names = [names[i] for i in order]
        scores = scores[order]

        plt.figure(figsize=(12, max(4, 0.35 * len(names))), dpi=self.cfg.fig_dpi)
        plt.barh(names, scores)
        plt.title("Fused Feature Scores (Novel Fusion)", fontsize=15, fontweight="bold", pad=10)
        plt.xlabel("Score", fontsize=12, fontweight="bold")
        plt.ylabel("Feature", fontsize=12, fontweight="bold")
        plt.tight_layout()
        plt.savefig(save_path, dpi=300, bbox_inches="tight")
        if show_plot:
            plt.show()
        else:
            plt.close()
        print(f"Saved fused score plot: {save_path}")


# ==========================
# MAIN EXECUTION
# ==========================
def main():
    df = pd.read_csv("train_80.csv")
    if "loan_status" not in df.columns:
        raise ValueError("Expected target column named 'loan_status' in train_80.csv")

    X_df = df.drop(columns=["loan_status"])
    y_raw = df["loan_status"].values

    le = LabelEncoder()
    y = le.fit_transform(y_raw)

    feature_names = X_df.columns.tolist()

    scaler = StandardScaler()
    X = scaler.fit_transform(X_df.values)

    cfg = HybridFSConfig(
        min_features=8,
        max_features=18,
        ga_generations=12,
        ga_pop_size=24,
        stability_bootstraps=30,
        perm_repeats=5,
        heatmap_cmap="viridis"
    )

    selector = ICMLHybridFeatureSelector(cfg=cfg, random_state=42)
    X_sel = selector.fit_transform(X, y, feature_names)

    def names(idxs): return [feature_names[i] for i in sorted(idxs)]

    print("\n==================== Method-wise Top Features ====================")
    print("Embedded (top):", names(selector.emb_selected_))
    print("AutoML-PFI (top):", names(selector.aml_selected_))
    print("GA (top):", names(selector.ga_selected_))

    print("\n==================== Final Selected (Novel Fusion) ====================")
    final_names = names(selector.selected_)
    print(final_names)
    print(f"Selected {len(final_names)} / {len(feature_names)} features.")

    print("\n==================== AutoML Model Weights (Reliability) ====================")
    for k, v in selector.model_weights_.items():
        if k.startswith("AutoML_"):
            print(f"{k}: {v:.4f}")
    print(f"Fusion threshold τ (stored): {selector.model_weights_.get('Fusion_tau', None)}")

    selector.plot_heatmap(save_path="hybrid_feature_heatmap.png", show_plot=True)
    selector.plot_fused_bar(save_path="fused_feature_scores.png", show_plot=True)

    out = pd.DataFrame(X_sel, columns=final_names)
    out["loan_status"] = y_raw
    out.to_csv("train_80_selected_hybrid.csv", index=False)
    print("\nSaved selected dataset: train_80_selected_hybrid.csv")


if __name__ == "__main__":
    main()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import StratifiedKFold, RandomizedSearchCV
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score, classification_report,
                             confusion_matrix, roc_curve, auc)
import joblib
from collections import Counter

# === Assume X_train, y_train, X_test, y_test are defined ===

param_grid = {
    'n_estimators': [50, 100, 150],
    'max_depth': [6, 12, 18],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'bootstrap': [True, False]
}

skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)

fold_metrics = []
fold_conf_matrices = []
tprs = []
roc_aucs = []
mean_fpr = np.linspace(0, 1, 100)
best_params_list = []

print("Starting 10-fold CV with RandomizedSearchCV tuning...\n")

for fold, (train_idx, val_idx) in enumerate(skf.split(X_train, y_train), 1):
    print(f"=== Fold {fold} ===")
    X_tr, X_val = X_train.iloc[train_idx], X_train.iloc[val_idx]
    y_tr, y_val = y_train.iloc[train_idx], y_train.iloc[val_idx]

    rf = RandomForestClassifier(random_state=42)

    rcv = RandomizedSearchCV(
        rf,
        param_distributions=param_grid,
        n_iter=15,
        scoring='accuracy',
        cv=3,
        n_jobs=-1,
        random_state=42,
        verbose=0
    )

    rcv.fit(X_tr, y_tr)
    best_rf = rcv.best_estimator_
    best_params_list.append(rcv.best_params_)

    y_pred = best_rf.predict(X_val)
    y_prob = best_rf.predict_proba(X_val)[:, 1]

    acc = accuracy_score(y_val, y_pred)
    pre = precision_score(y_val, y_pred, average='weighted', zero_division=0)
    rec = recall_score(y_val, y_pred, average='weighted', zero_division=0)
    f1 = f1_score(y_val, y_pred, average='weighted', zero_division=0)
    roc = roc_auc_score(y_val, y_prob)

    fold_metrics.append((acc, pre, rec, f1, roc))

    print(f"Best params: {rcv.best_params_}")
    print(f"Accuracy:  {acc:.4f}")
    print(f"Precision: {pre:.4f}")
    print(f"Recall:    {rec:.4f}")
    print(f"F1-score:  {f1:.4f}")
    print(f"ROC AUC:   {roc:.4f}\n")

    cm = confusion_matrix(y_val, y_pred)
    fold_conf_matrices.append(cm)

    fpr, tpr, _ = roc_curve(y_val, y_prob)
    tprs.append(np.interp(mean_fpr, fpr, tpr))
    tprs[-1][0] = 0.0
    roc_aucs.append(roc)

metrics_array = np.array(fold_metrics)
metrics_names = ['Accuracy', 'Precision', 'Recall', 'F1', 'ROC AUC']

print("=== Cross-validation Summary ===")
for i, metric_name in enumerate(metrics_names):
    mean_val = metrics_array[:, i].mean()
    std_val = metrics_array[:, i].std()
    print(f"{metric_name}: Mean = {mean_val:.4f}, Std = {std_val:.4f}")

sum_cm = np.sum(fold_conf_matrices, axis=0)

# CV Confusion Matrix (Greens)
plt.figure(figsize=(8, 8))
sns.heatmap(sum_cm, annot=True, fmt='d', cmap='Greens', cbar=False,
            annot_kws={"size": 20})
plt.title('Aggregated Confusion Matrix (CV)', fontsize=24, fontweight='bold')
plt.xlabel('Predicted Label', fontsize=18)
plt.ylabel('True Label', fontsize=18)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.tight_layout()
plt.show()

mean_tpr = np.mean(tprs, axis=0)
mean_tpr[-1] = 1.0
mean_auc = auc(mean_fpr, mean_tpr)
std_auc = np.std(roc_aucs)

tprs_upper = np.minimum(mean_tpr + np.std(tprs, axis=0), 1)
tprs_lower = np.maximum(mean_tpr - np.std(tprs, axis=0), 0)

plt.figure(figsize=(10, 8))
plt.plot(mean_fpr, mean_tpr, color='blue', lw=4,
         label=f'Mean CV ROC (AUC = {mean_auc:.4f} ± {std_auc:.4f})')
plt.fill_between(mean_fpr, tprs_lower, tprs_upper, color='blue', alpha=0.2)
plt.plot([0, 1], [0, 1], 'k--', lw=2)

plt.xlabel('False Positive Rate', fontsize=20)
plt.ylabel('True Positive Rate', fontsize=20)
plt.title('Mean ROC Curve with ±1 Std Dev (CV)', fontsize=24, fontweight='bold')
plt.legend(loc='lower right', fontsize=16)
plt.grid(alpha=0.7, linestyle='--')
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.tight_layout()
plt.show()

most_common_params = Counter(tuple(sorted(p.items())) for p in best_params_list).most_common(1)[0][0]
final_params = dict(most_common_params)

print("\nFinal hyperparameters selected:")
print(final_params)

final_model = RandomForestClassifier(random_state=42, **final_params)
final_model.fit(X_train, y_train)

y_test_pred = final_model.predict(X_test)
y_test_prob = final_model.predict_proba(X_test)[:, 1]

acc = accuracy_score(y_test, y_test_pred)
pre = precision_score(y_test, y_test_pred, average='weighted', zero_division=0)
rec = recall_score(y_test, y_test_pred, average='weighted', zero_division=0)
f1 = f1_score(y_test, y_test_pred, average='weighted', zero_division=0)
roc_auc = roc_auc_score(y_test, y_test_prob)

print("\n=== Test Set Performance ===")
print(classification_report(y_test, y_test_pred))
print(f"Accuracy:  {acc:.4f}")
print(f"Precision: {pre:.4f}")
print(f"Recall:    {rec:.4f}")
print(f"F1-Score:  {f1:.4f}")
print(f"ROC-AUC:   {roc_auc:.4f}")

cm_test = confusion_matrix(y_test, y_test_pred)
plt.figure(figsize=(8, 8))
sns.heatmap(cm_test, annot=True, fmt='d', cmap='Greens', cbar=False,
            annot_kws={"size": 20})
plt.title('Test Set Confusion Matrix', fontsize=24, fontweight='bold')
plt.xlabel('Predicted Label', fontsize=18)
plt.ylabel('True Label', fontsize=18)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.tight_layout()
plt.show()

# === 1) Combined ROC: Mean CV + Test ROC ===
plt.figure(figsize=(10, 8))
plt.plot(mean_fpr, mean_tpr, color='blue', lw=4,
         label=f'Mean CV ROC (AUC = {mean_auc:.4f} ± {std_auc:.4f})')
plt.fill_between(mean_fpr, tprs_lower, tprs_upper, color='blue', alpha=0.2)

fpr_test, tpr_test, _ = roc_curve(y_test, y_test_prob)
plt.plot(fpr_test, tpr_test, color='red', lw=4,
         label=f'Test ROC (AUC = {roc_auc:.4f})')

plt.plot([0, 1], [0, 1], 'k--', lw=2)
plt.xlabel('False Positive Rate', fontsize=20)
plt.ylabel('True Positive Rate', fontsize=20)
plt.title('ROC Curve: CV vs Test', fontsize=24, fontweight='bold')
plt.legend(loc='lower right', fontsize=16)
plt.grid(alpha=0.7, linestyle='--')
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.tight_layout()
plt.show()

# === 2) Separate Test ROC with same smooth design ===
# Interpolate test ROC for smoothness
mean_fpr_test = np.linspace(0, 1, 100)
interp_tpr_test = np.interp(mean_fpr_test, fpr_test, tpr_test)
interp_tpr_test[0] = 0.0
interp_tpr_test[-1] = 1.0

plt.figure(figsize=(10, 8))
plt.plot(mean_fpr_test, interp_tpr_test, color='red', lw=4,
         label=f'Test ROC (AUC = {roc_auc:.4f})')

# Add a narrow shaded band for style consistency
plt.fill_between(mean_fpr_test,
                 np.maximum(interp_tpr_test - 0.01, 0),
                 np.minimum(interp_tpr_test + 0.01, 1),
                 color='red', alpha=0.2)

plt.plot([0, 1], [0, 1], 'k--', lw=2)
plt.xlabel('False Positive Rate', fontsize=20)
plt.ylabel('True Positive Rate', fontsize=20)
plt.title('ROC Curve: Test Set Only', fontsize=24, fontweight='bold')
plt.legend(loc='lower right', fontsize=16)
plt.grid(alpha=0.7, linestyle='--')
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.tight_layout()
plt.show()

# Save model
joblib.dump(final_model, 'random_forest_tuned_10fold.pkl')
print("\n✅ Final model saved as 'random_forest_tuned_10fold.pkl'")
```

# **RF ETC**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier, StackingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold, RandomizedSearchCV
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score, classification_report,
                             confusion_matrix, roc_curve, auc)
import joblib
from collections import Counter

# === Assume X_train, y_train, X_test, y_test are defined ===

# Hyperparameter grids for base models
param_grid_rf = {
    'n_estimators': [50, 100, 150],
    'max_depth': [6, 12, 18],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'bootstrap': [True, False]
}

param_grid_etc = {
    'n_estimators': [50, 100, 150],
    'max_depth': [6, 12, 18],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'bootstrap': [True, False]
}

# Hyperparameter grid for meta model (LogisticRegression)
param_grid_meta = {
    'C': [0.01, 0.1, 1],
    'penalty': ['l1', 'l2', 'elasticnet'],
    'solver': ['saga', 'liblinear', 'saga']  # saga twice to allow elasticnet penalty
}

# Function to tune a base estimator with RandomizedSearchCV on training fold
def tune_base_model(model, param_grid, X_tr, y_tr):
    rcv = RandomizedSearchCV(
        model,
        param_distributions=param_grid,
        n_iter=15,
        scoring='accuracy',
        cv=3,
        n_jobs=-1,
        random_state=42,
        verbose=0
    )
    rcv.fit(X_tr, y_tr)
    return rcv.best_estimator_, rcv.best_params_

# Tune meta model within stacking, so param tuning on meta will be done later with RandomizedSearchCV

skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)

fold_metrics = []
fold_conf_matrices = []
tprs = []
roc_aucs = []
mean_fpr = np.linspace(0, 1, 100)

print("Starting 10-fold CV for stacking ensemble...\n")

for fold, (train_idx, val_idx) in enumerate(skf.split(X_train, y_train), 1):
    print(f"=== Fold {fold} ===")
    X_tr, X_val = X_train.iloc[train_idx], X_train.iloc[val_idx]
    y_tr, y_val = y_train.iloc[train_idx], y_train.iloc[val_idx]

    # Tune base estimators on training fold
    print("Tuning RandomForest...")
    best_rf, best_rf_params = tune_base_model(RandomForestClassifier(random_state=42), param_grid_rf, X_tr, y_tr)
    print(f"RF best params: {best_rf_params}")

    print("Tuning ExtraTrees...")
    best_etc, best_etc_params = tune_base_model(ExtraTreesClassifier(random_state=42), param_grid_etc, X_tr, y_tr)
    print(f"ETC best params: {best_etc_params}")

    # Define stacking with tuned base estimators (frozen params)
    base_estimators = [
        ('rf', best_rf),
        ('etc', best_etc)
    ]

    # Meta model — Logistic Regression (tune separately on fold training)
    meta_lr = LogisticRegression(random_state=42, max_iter=1000)

    # Define stacking classifier
    stacking_clf = StackingClassifier(
        estimators=base_estimators,
        final_estimator=meta_lr,
        cv=3,
        n_jobs=-1,
        passthrough=False,
        verbose=0
    )

    # Tune meta model parameters using RandomizedSearchCV on X_tr,y_tr with stacking fixed base
    print("Tuning meta LogisticRegression inside stacking...")
    meta_rcv = RandomizedSearchCV(
        stacking_clf,
        param_distributions={
            'final_estimator__C': param_grid_meta['C'],
            'final_estimator__penalty': param_grid_meta['penalty'],
            'final_estimator__solver': param_grid_meta['solver'],
        },
        n_iter=15,
        scoring='accuracy',
        cv=3,
        n_jobs=-1,
        random_state=42,
        verbose=0
    )

    meta_rcv.fit(X_tr, y_tr)
    best_stacking = meta_rcv.best_estimator_

    print(f"Best meta params: {meta_rcv.best_params_}")

    # Predict on validation fold
    y_pred = best_stacking.predict(X_val)
    y_prob = best_stacking.predict_proba(X_val)[:, 1]

    acc = accuracy_score(y_val, y_pred)
    pre = precision_score(y_val, y_pred, average='weighted', zero_division=0)
    rec = recall_score(y_val, y_pred, average='weighted', zero_division=0)
    f1 = f1_score(y_val, y_pred, average='weighted', zero_division=0)
    roc = roc_auc_score(y_val, y_prob)

    fold_metrics.append((acc, pre, rec, f1, roc))

    print(f"Accuracy:  {acc:.4f}")
    print(f"Precision: {pre:.4f}")
    print(f"Recall:    {rec:.4f}")
    print(f"F1-score:  {f1:.4f}")
    print(f"ROC AUC:   {roc:.4f}\n")

    cm = confusion_matrix(y_val, y_pred)
    fold_conf_matrices.append(cm)

    fpr, tpr, _ = roc_curve(y_val, y_prob)
    tprs.append(np.interp(mean_fpr, fpr, tpr))
    tprs[-1][0] = 0.0
    roc_aucs.append(roc)

# CV metrics summary
metrics_array = np.array(fold_metrics)
metrics_names = ['Accuracy', 'Precision', 'Recall', 'F1', 'ROC AUC']

print("=== Cross-validation Summary ===")
for i, metric_name in enumerate(metrics_names):
    mean_val = metrics_array[:, i].mean()
    std_val = metrics_array[:, i].std()
    print(f"{metric_name}: Mean = {mean_val:.4f}, Std = {std_val:.4f}")

sum_cm = np.sum(fold_conf_matrices, axis=0)

plt.figure(figsize=(8, 8))
sns.heatmap(sum_cm, annot=True, fmt='d', cmap='Greens', cbar=False,
            annot_kws={"size": 20})
plt.title('Aggregated Confusion Matrix (CV)', fontsize=24, fontweight='bold')
plt.xlabel('Predicted Label', fontsize=18)
plt.ylabel('True Label', fontsize=18)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.tight_layout()
plt.show()

mean_tpr = np.mean(tprs, axis=0)
mean_tpr[-1] = 1.0
mean_auc = auc(mean_fpr, mean_tpr)
std_auc = np.std(roc_aucs)

tprs_upper = np.minimum(mean_tpr + np.std(tprs, axis=0), 1)
tprs_lower = np.maximum(mean_tpr - np.std(tprs, axis=0), 0)

plt.figure(figsize=(10, 8))
plt.plot(mean_fpr, mean_tpr, color='blue', lw=4,
         label=f'Mean CV ROC (AUC = {mean_auc:.4f} ± {std_auc:.4f})')
plt.fill_between(mean_fpr, tprs_lower, tprs_upper, color='blue', alpha=0.2)
plt.plot([0, 1], [0, 1], 'k--', lw=2)

plt.xlabel('False Positive Rate', fontsize=20)
plt.ylabel('True Positive Rate', fontsize=20)
plt.title('Mean ROC Curve with ±1 Std Dev (CV)', fontsize=24, fontweight='bold')
plt.legend(loc='lower right', fontsize=16)
plt.grid(alpha=0.7, linestyle='--')
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.tight_layout()
plt.show()

# Final model training on full data:

print("\nTraining final stacking model on full training data...")

# Tune base estimators on full training set
final_rf, final_rf_params = tune_base_model(RandomForestClassifier(random_state=42), param_grid_rf, X_train, y_train)
final_etc, final_etc_params = tune_base_model(ExtraTreesClassifier(random_state=42), param_grid_etc, X_train, y_train)

print(f"Final RF params: {final_rf_params}")
print(f"Final ETC params: {final_etc_params}")

base_estimators_final = [
    ('rf', final_rf),
    ('etc', final_etc)
]

# Stacking with unfitted meta (will tune meta next)
stacking_final = StackingClassifier(
    estimators=base_estimators_final,
    final_estimator=LogisticRegression(random_state=42, max_iter=1000),
    cv=3,
    n_jobs=-1,
    passthrough=False,
    verbose=0
)

# Tune meta parameters on full training data
meta_rcv_final = RandomizedSearchCV(
    stacking_final,
    param_distributions={
        'final_estimator__C': param_grid_meta['C'],
        'final_estimator__penalty': param_grid_meta['penalty'],
        'final_estimator__solver': param_grid_meta['solver'],
    },
    n_iter=15,
    scoring='accuracy',
    cv=3,
    n_jobs=-1,
    random_state=42,
    verbose=0
)

meta_rcv_final.fit(X_train, y_train)
final_model = meta_rcv_final.best_estimator_

print(f"Final meta params: {meta_rcv_final.best_params_}")

# Test set evaluation
y_test_pred = final_model.predict(X_test)
y_test_prob = final_model.predict_proba(X_test)[:, 1]

acc = accuracy_score(y_test, y_test_pred)
pre = precision_score(y_test, y_test_pred, average='weighted', zero_division=0)
rec = recall_score(y_test, y_test_pred, average='weighted', zero_division=0)
f1 = f1_score(y_test, y_test_pred, average='weighted', zero_division=0)
roc_auc = roc_auc_score(y_test, y_test_prob)

print("\n=== Test Set Performance ===")
print(classification_report(y_test, y_test_pred))
print(f"Accuracy:  {acc:.4f}")
print(f"Precision: {pre:.4f}")
print(f"Recall:    {rec:.4f}")
print(f"F1-score:  {f1:.4f}")
print(f"ROC AUC:   {roc_auc:.4f}")

cm_test = confusion_matrix(y_test, y_test_pred)
plt.figure(figsize=(8, 8))
sns.heatmap(cm_test, annot=True, fmt='d', cmap='Greens', cbar=False,
            annot_kws={"size": 20})
plt.title('Test Set Confusion Matrix', fontsize=24, fontweight='bold')
plt.xlabel('Predicted Label', fontsize=18)
plt.ylabel('True Label', fontsize=18)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.tight_layout()
plt.show()

# Plot ROC curves for CV mean and test (same style)

plt.figure(figsize=(10, 8))
plt.plot(mean_fpr, mean_tpr, color='blue', lw=4,
         label=f'Mean CV ROC (AUC = {mean_auc:.4f} ± {std_auc:.4f})')
plt.fill_between(mean_fpr, tprs_lower, tprs_upper, color='blue', alpha=0.2)

fpr_test, tpr_test, _ = roc_curve(y_test, y_test_prob)
plt.plot(fpr_test, tpr_test, color='red', lw=4,
         label=f'Test ROC (AUC = {roc_auc:.4f})')

plt.plot([0, 1], [0, 1], 'k--', lw=2)
plt.xlabel('False Positive Rate', fontsize=20)
plt.ylabel('True Positive Rate', fontsize=20)
plt.title('ROC Curve: CV vs Test', fontsize=24, fontweight='bold')
plt.legend(loc='lower right', fontsize=16)
plt.grid(alpha=0.7, linestyle='--')
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.tight_layout()
plt.show()

# Separate Test ROC plot with same style
mean_fpr_test = np.linspace(0, 1, 100)
interp_tpr_test = np.interp(mean_fpr_test, fpr_test, tpr_test)
interp_tpr_test[0] = 0.0
interp_tpr_test[-1] = 1.0

plt.figure(figsize=(10, 8))
plt.plot(mean_fpr_test, interp_tpr_test, color='red', lw=4,
         label=f'Test ROC (AUC = {roc_auc:.4f})')

plt.fill_between(mean_fpr_test,
                 np.maximum(interp_tpr_test - 0.01, 0),
                 np.minimum(interp_tpr_test + 0.01, 1),
                 color='red', alpha=0.2)

plt.plot([0, 1], [0, 1], 'k--', lw=2)
plt.xlabel('False Positive Rate', fontsize=20)
plt.ylabel('True Positive Rate', fontsize=20)
plt.title('ROC Curve: Test Set Only', fontsize=24, fontweight='bold')
plt.legend(loc='lower right', fontsize=16)
plt.grid(alpha=0.7, linestyle='--')
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.tight_layout()
plt.show()

# Save final stacking model
joblib.dump(final_model, 'stacking_ensemble_rf_etc_lr.pkl')
print("\n✅ Final stacking ensemble model saved as 'stacking_ensemble_rf_etc_lr.pkl'")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.ensemble import RandomForestClassifier, StackingClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold, RandomizedSearchCV
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score, classification_report,
                             confusion_matrix, roc_curve, auc)
import joblib
from collections import Counter

# === Assume X_train, y_train, X_test, y_test are defined ===

param_grid_knn = {
    'n_neighbors': [3, 5, 7],
    'weights': ['uniform', 'distance', 'uniform'],
    'algorithm': ['auto', 'ball_tree', 'kd_tree'],
    'leaf_size': [20, 30, 40],
    'p': [1, 2, 3]
}

param_grid_rf = {
    'n_estimators': [50, 100, 150],
    'max_depth': [6, 12, 18],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'bootstrap': [True, False]
}

param_grid_meta = {
    'C': [0.01, 0.1, 1],
    'penalty': ['l1', 'l2', 'elasticnet'],
    'solver': ['saga', 'liblinear', 'saga']
}

def tune_base_model(model, param_grid, X_tr, y_tr):
    rcv = RandomizedSearchCV(
        model,
        param_distributions=param_grid,
        n_iter=15,
        scoring='accuracy',
        cv=3,
        n_jobs=-1,
        random_state=42,
        verbose=0
    )
    rcv.fit(X_tr, y_tr)
    return rcv.best_estimator_, rcv.best_params_

skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)

fold_metrics = []
fold_conf_matrices = []
tprs = []
roc_aucs = []
mean_fpr = np.linspace(0, 1, 100)

print("Starting 10-fold CV for stacking ensemble with KNN + RF...\n")

for fold, (train_idx, val_idx) in enumerate(skf.split(X_train, y_train), 1):
    print(f"=== Fold {fold} ===")
    X_tr, X_val = X_train.iloc[train_idx], X_train.iloc[val_idx]
    y_tr, y_val = y_train.iloc[train_idx], y_train.iloc[val_idx]

    print("Tuning KNN...")
    best_knn, best_knn_params = tune_base_model(KNeighborsClassifier(), param_grid_knn, X_tr, y_tr)
    print(f"KNN best params: {best_knn_params}")

    print("Tuning RandomForest...")
    best_rf, best_rf_params = tune_base_model(RandomForestClassifier(random_state=42), param_grid_rf, X_tr, y_tr)
    print(f"RF best params: {best_rf_params}")

    base_estimators = [
        ('knn', best_knn),
        ('rf', best_rf)
    ]

    meta_lr = LogisticRegression(random_state=42, max_iter=1000)

    stacking_clf = StackingClassifier(
        estimators=base_estimators,
        final_estimator=meta_lr,
        cv=3,
        n_jobs=-1,
        passthrough=False,
        verbose=0
    )

    print("Tuning meta LogisticRegression inside stacking...")
    meta_rcv = RandomizedSearchCV(
        stacking_clf,
        param_distributions={
            'final_estimator__C': param_grid_meta['C'],
            'final_estimator__penalty': param_grid_meta['penalty'],
            'final_estimator__solver': param_grid_meta['solver'],
        },
        n_iter=15,
        scoring='accuracy',
        cv=3,
        n_jobs=-1,
        random_state=42,
        verbose=0
    )
    meta_rcv.fit(X_tr, y_tr)
    best_stacking = meta_rcv.best_estimator_

    print(f"Best meta params: {meta_rcv.best_params_}")

    y_pred = best_stacking.predict(X_val)
    y_prob = best_stacking.predict_proba(X_val)[:, 1]

    acc = accuracy_score(y_val, y_pred)
    pre = precision_score(y_val, y_pred, average='weighted', zero_division=0)
    rec = recall_score(y_val, y_pred, average='weighted', zero_division=0)
    f1 = f1_score(y_val, y_pred, average='weighted', zero_division=0)
    roc = roc_auc_score(y_val, y_prob)

    fold_metrics.append((acc, pre, rec, f1, roc))

    print(f"Accuracy:  {acc:.4f}")
    print(f"Precision: {pre:.4f}")
    print(f"Recall:    {rec:.4f}")
    print(f"F1-score:  {f1:.4f}")
    print(f"ROC AUC:   {roc:.4f}\n")

    cm = confusion_matrix(y_val, y_pred)
    fold_conf_matrices.append(cm)

    fpr, tpr, _ = roc_curve(y_val, y_prob)
    tprs.append(np.interp(mean_fpr, fpr, tpr))
    tprs[-1][0] = 0.0
    roc_aucs.append(roc)

metrics_array = np.array(fold_metrics)
metrics_names = ['Accuracy', 'Precision', 'Recall', 'F1', 'ROC AUC']

print("=== Cross-validation Summary ===")
for i, metric_name in enumerate(metrics_names):
    mean_val = metrics_array[:, i].mean()
    std_val = metrics_array[:, i].std()
    print(f"{metric_name}: Mean = {mean_val:.4f}, Std = {std_val:.4f}")

sum_cm = np.sum(fold_conf_matrices, axis=0)

plt.figure(figsize=(8, 8))
sns.heatmap(sum_cm, annot=True, fmt='d', cmap='Greens', cbar=False,
            annot_kws={"size": 20})
plt.title('Aggregated Confusion Matrix (CV)', fontsize=24, fontweight='bold')
plt.xlabel('Predicted Label', fontsize=18)
plt.ylabel('True Label', fontsize=18)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.tight_layout()
plt.show()

mean_tpr = np.mean(tprs, axis=0)
mean_tpr[-1] = 1.0
mean_auc = auc(mean_fpr, mean_tpr)
std_auc = np.std(roc_aucs)

tprs_upper = np.minimum(mean_tpr + np.std(tprs, axis=0), 1)
tprs_lower = np.maximum(mean_tpr - np.std(tprs, axis=0), 0)

plt.figure(figsize=(10, 8))
plt.plot(mean_fpr, mean_tpr, color='blue', lw=4,
         label=f'Mean CV ROC (AUC = {mean_auc:.4f} ± {std_auc:.4f})')
plt.fill_between(mean_fpr, tprs_lower, tprs_upper, color='blue', alpha=0.2)
plt.plot([0, 1], [0, 1], 'k--', lw=2)

plt.xlabel('False Positive Rate', fontsize=20)
plt.ylabel('True Positive Rate', fontsize=20)
plt.title('Mean ROC Curve with ±1 Std Dev (CV)', fontsize=24, fontweight='bold')
plt.legend(loc='lower right', fontsize=16)
plt.grid(alpha=0.7, linestyle='--')
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.tight_layout()
plt.show()

# Final model training on full data

print("\nTraining final stacking model on full training data...")

final_knn, final_knn_params = tune_base_model(KNeighborsClassifier(), param_grid_knn, X_train, y_train)
final_rf, final_rf_params = tune_base_model(RandomForestClassifier(random_state=42), param_grid_rf, X_train, y_train)

print(f"Final KNN params: {final_knn_params}")
print(f"Final RF params: {final_rf_params}")

base_estimators_final = [
    ('knn', final_knn),
    ('rf', final_rf)
]

stacking_final = StackingClassifier(
    estimators=base_estimators_final,
    final_estimator=LogisticRegression(random_state=42, max_iter=1000),
    cv=3,
    n_jobs=-1,
    passthrough=False,
    verbose=0
)

meta_rcv_final = RandomizedSearchCV(
    stacking_final,
    param_distributions={
        'final_estimator__C': param_grid_meta['C'],
        'final_estimator__penalty': param_grid_meta['penalty'],
        'final_estimator__solver': param_grid_meta['solver'],
    },
    n_iter=15,
    scoring='accuracy',
    cv=3,
    n_jobs=-1,
    random_state=42,
    verbose=0
)

meta_rcv_final.fit(X_train, y_train)
final_model = meta_rcv_final.best_estimator_

print(f"Final meta params: {meta_rcv_final.best_params_}")

# Test evaluation

y_test_pred = final_model.predict(X_test)
y_test_prob = final_model.predict_proba(X_test)[:, 1]

acc = accuracy_score(y_test, y_test_pred)
pre = precision_score(y_test, y_test_pred, average='weighted', zero_division=0)
rec = recall_score(y_test, y_test_pred, average='weighted', zero_division=0)
f1 = f1_score(y_test, y_test_pred, average='weighted', zero_division=0)
roc_auc = roc_auc_score(y_test, y_test_prob)

print("\n=== Test Set Performance ===")
print(classification_report(y_test, y_test_pred))
print(f"Accuracy:  {acc:.4f}")
print(f"Precision: {pre:.4f}")
print(f"Recall:    {rec:.4f}")
print(f"F1-score:  {f1:.4f}")
print(f"ROC AUC:   {roc_auc:.4f}")

cm_test = confusion_matrix(y_test, y_test_pred)
plt.figure(figsize=(8, 8))
sns.heatmap(cm_test, annot=True, fmt='d', cmap='Greens', cbar=False,
            annot_kws={"size": 20})
plt.title('Test Set Confusion Matrix', fontsize=24, fontweight='bold')
plt.xlabel('Predicted Label', fontsize=18)
plt.ylabel('True Label', fontsize=18)
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 8))
plt.plot(mean_fpr, mean_tpr, color='blue', lw=4,
         label=f'Mean CV ROC (AUC = {mean_auc:.4f} ± {std_auc:.4f})')
plt.fill_between(mean_fpr, tprs_lower, tprs_upper, color='blue', alpha=0.2)

fpr_test, tpr_test, _ = roc_curve(y_test, y_test_prob)
plt.plot(fpr_test, tpr_test, color='red', lw=4,
         label=f'Test ROC (AUC = {roc_auc:.4f})')

plt.plot([0, 1], [0, 1], 'k--', lw=2)
plt.xlabel('False Positive Rate', fontsize=20)
plt.ylabel('True Positive Rate', fontsize=20)
plt.title('ROC Curve: CV vs Test', fontsize=24, fontweight='bold')
plt.legend(loc='lower right', fontsize=16)
plt.grid(alpha=0.7, linestyle='--')
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.tight_layout()
plt.show()

mean_fpr_test = np.linspace(0, 1, 100)
interp_tpr_test = np.interp(mean_fpr_test, fpr_test, tpr_test)
interp_tpr_test[0] = 0.0
interp_tpr_test[-1] = 1.0

plt.figure(figsize=(10, 8))
plt.plot(mean_fpr_test, interp_tpr_test, color='red', lw=4,
         label=f'Test ROC (AUC = {roc_auc:.4f})')

plt.fill_between(mean_fpr_test,
                 np.maximum(interp_tpr_test - 0.01, 0),
                 np.minimum(interp_tpr_test + 0.01, 1),
                 color='red', alpha=0.2)

plt.plot([0, 1], [0, 1], 'k--', lw=2)
plt.xlabel('False Positive Rate', fontsize=20)
plt.ylabel('True Positive Rate', fontsize=20)
plt.title('ROC Curve: Test Set Only', fontsize=24, fontweight='bold')
plt.legend(loc='lower right', fontsize=16)
plt.grid(alpha=0.7, linestyle='--')
plt.xticks(fontsize=16)
plt.yticks(fontsize=16)
plt.tight_layout()
plt.show()

joblib.dump(final_model, 'stacking_ensemble_knn_rf_lr.pkl')
print("\n✅ Final stacking ensemble model saved as 'stacking_ensemble_knn_rf_lr.pkl'")
```

```python
# =========================
# XAI for saved sklearn model (Colab-friendly)
# Fixes:
#  - Feature-name mismatch (align columns to model.feature_names_in_)
#  - Avoids breaking Colab by not upgrading pandas past the pinned version
# Files expected in Colab working dir:
#   - stacking_ensemble_knn_rf_lr.pkl
#   - test_20.csv
# Target column (if present): loan_status
# =========================

# --- Install (Colab-safe) ---
# NOTE: Colab pins pandas==2.2.2, so DON'T upgrade pandas.
# Also: you don't need alibi here; it conflicts with NumPy>=2 in many setups.
!pip -q install -U "numpy>=2.0" "scikit-learn>=1.4" shap lime matplotlib joblib

import os
import joblib
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt

from sklearn.inspection import permutation_importance, PartialDependenceDisplay
from sklearn.metrics import ConfusionMatrixDisplay, RocCurveDisplay, PrecisionRecallDisplay
from sklearn.calibration import calibration_curve  # ✅ correct import

# --- Global plot style ---
mpl.rcParams.update({
    "figure.dpi": 180,
    "savefig.dpi": 300,
    "font.size": 12,
    "axes.titlesize": 14,
    "axes.labelsize": 12,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11,
    "legend.fontsize": 11,
    "axes.grid": True,
    "grid.alpha": 0.25
})

# -------------------------
# 1) Load model + data
# -------------------------
MODEL_PATH = "stacking_ensemble_knn_rf_lr.pkl"
TEST_PATH  = "test_20.csv"
TARGET_COL = "loan_status"

assert os.path.exists(MODEL_PATH), f"Missing: {MODEL_PATH}"
assert os.path.exists(TEST_PATH),  f"Missing: {TEST_PATH}"

model = joblib.load(MODEL_PATH)
df = pd.read_csv(TEST_PATH)

if TARGET_COL in df.columns:
    y = df[TARGET_COL].copy()
    X = df.drop(columns=[TARGET_COL]).copy()
else:
    y = None
    X = df.copy()

print("Model loaded:", type(model))
print("Raw X shape:", X.shape, "| y present:", y is not None)

# -------------------------
# 2) Helpers (feature alignment + proba)
# -------------------------
def _get_estimator_for_classes(m):
    """Return an object that likely exposes classes_."""
    if hasattr(m, "classes_"):
        return m
    if hasattr(m, "named_steps"):  # Pipeline
        for key in ["classifier", "model"]:
            if key in m.named_steps and hasattr(m.named_steps[key], "classes_"):
                return m.named_steps[key]
        last = list(m.named_steps.values())[-1]
        if hasattr(last, "classes_"):
            return last
    return None

def get_pos_index(m):
    """Choose positive class index: prefer class 1 if present, else last."""
    est = _get_estimator_for_classes(m)
    if est is None:
        return 1  # assume binary [:,1]
    classes = list(est.classes_)
    if 1 in classes:
        return classes.index(1)
    if "1" in classes:
        return classes.index("1")
    return min(1, len(classes) - 1)

def _get_feature_names_in(m):
    """
    Best-effort: find feature_names_in_ from model or pipeline steps.
    Works for many sklearn estimators, including StackingClassifier fit on a DataFrame.
    """
    if hasattr(m, "feature_names_in_"):
        return list(m.feature_names_in_)
    if hasattr(m, "named_steps"):  # pipeline
        # search from last to first
        for step in reversed(list(m.named_steps.values())):
            if hasattr(step, "feature_names_in_"):
                return list(step.feature_names_in_)
    # try stacked estimators (sometimes stored there)
    if hasattr(m, "estimators_"):
        for est in m.estimators_:
            if hasattr(est, "feature_names_in_"):
                return list(est.feature_names_in_)
    return None

def align_to_model_features(X_df, m, fill_value=0.0):
    """
    Align X_df columns to the model's training columns:
      - drop unseen columns
      - add missing columns (filled with fill_value)
      - reorder to match training order
    If model does not expose feature_names_in_, returns X_df unchanged.
    """
    expected = _get_feature_names_in(m)
    if expected is None:
        print("⚠️ Model has no feature_names_in_. Skipping alignment (assumes correct order already).")
        return X_df

    expected_set = set(expected)
    got_set = set(X_df.columns)

    extra = sorted(list(got_set - expected_set))
    missing = sorted(list(expected_set - got_set))

    if extra:
        print(f"🧹 Dropping {len(extra)} unseen columns (not used during fit). Example:", extra[:8])
        X_df = X_df.drop(columns=extra)

    if missing:
        print(f"➕ Adding {len(missing)} missing columns (were used during fit). Example:", missing[:8])
        for c in missing:
            X_df[c] = fill_value

    # reorder
    X_df = X_df[expected].copy()
    print("✅ Aligned X shape:", X_df.shape)
    return X_df

def predict_proba_fn(data):
    """Works for most sklearn classifiers & pipelines; returns proba (n, K)."""
    if hasattr(model, "predict_proba"):
        return model.predict_proba(data)

    # fallback: decision_function -> sigmoid for binary
    if hasattr(model, "decision_function"):
        s = model.decision_function(data)
        s = np.asarray(s)
        if s.ndim == 1:  # binary
            p1 = 1.0 / (1.0 + np.exp(-s))
            return np.column_stack([1 - p1, p1])

    raise ValueError("Model does not support predict_proba or compatible decision_function.")

# ---- FIX FEATURE NAME MISMATCH HERE ----
X = align_to_model_features(X, model, fill_value=0.0)

pos_idx = get_pos_index(model)
print("Positive class index:", pos_idx)

# Quick numeric sanity check (many stacking models require purely numeric input)
non_numeric = [c for c in X.columns if not np.issubdtype(X[c].dtype, np.number)]
if non_numeric:
    print("⚠️ Non-numeric columns detected (model may fail unless it was trained with them):", non_numeric[:10])

# subsamples for speed
X_bg   = X.sample(min(200, len(X)), random_state=42)   # background for SHAP/LIME
X_shap = X.sample(min(800, len(X)), random_state=7)    # points to explain globally
x0     = X.iloc[[0]].copy()                            # one instance

# -------------------------
# 3) SHAP (global + local)
# -------------------------
import shap
shap.initjs()

def f_posprob(data):
    return predict_proba_fn(data)[:, pos_idx]

# Use Independent masker on aligned data
masker = shap.maskers.Independent(X_bg)

# For arbitrary sklearn models, PermutationExplainer is robust (slower, but safe).
# We cap evaluations for speed/stability.
explainer = shap.Explainer(f_posprob, masker, algorithm="permutation")
shap_values = explainer(X_shap, max_evals=2 * X_shap.shape[1] + 1)

# Plot 1: SHAP beeswarm
plt.figure(figsize=(11, 6))
shap.plots.beeswarm(shap_values, max_display=20, show=False)
plt.title("SHAP Summary (Beeswarm) - Global", pad=12)
plt.tight_layout()
plt.show()

# Plot 2: SHAP bar
plt.figure(figsize=(11, 6))
shap.plots.bar(shap_values, max_display=20, show=False)
plt.title("SHAP Feature Importance (Bar) - Global", pad=12)
plt.tight_layout()
plt.show()

# Top features
mean_abs = np.abs(shap_values.values).mean(axis=0)
top_idx = np.argsort(mean_abs)[::-1]
top_features = [X_shap.columns[i] for i in top_idx[:6]]
print("Top features (by mean |SHAP|):", top_features)

# Plot 3-5: Dependence plots (top 3)
for feat in top_features[:3]:
    plt.figure(figsize=(9.5, 6))
    shap.dependence_plot(feat, shap_values.values, X_shap, show=False)
    plt.title(f"SHAP Dependence - {feat}", pad=12)
    plt.tight_layout()
    plt.show()

# Plot 6: Interaction-ish dependence (color by 2nd top feature)
feat_a = top_features[0]
feat_b = top_features[1] if len(top_features) > 1 else top_features[0]
plt.figure(figsize=(9.5, 6))
shap.dependence_plot(feat_a, shap_values.values, X_shap, interaction_index=feat_b, show=False)
plt.title(f"SHAP Interaction (color={feat_b}) - {feat_a}", pad=12)
plt.tight_layout()
plt.show()

# Local SHAP for one instance
sv0 = explainer(x0, max_evals=2 * x0.shape[1] + 1)

# Plot 7: Waterfall
plt.figure(figsize=(10, 6))
shap.plots.waterfall(sv0[0], max_display=20, show=False)
plt.title("SHAP Waterfall - Local (first row)", pad=12)
plt.tight_layout()
plt.show()

# Plot 8: Force plot (interactive)
shap.force_plot(sv0[0].base_values, sv0[0].values, x0, matplotlib=False)

# Plot 9: Decision plot
plt.figure(figsize=(11, 5))
shap.decision_plot(sv0[0].base_values, sv0[0].values, x0.columns, show=False)
plt.title("SHAP Decision Plot - Local (first row)", pad=12)
plt.tight_layout()
plt.show()

# Plot 10: Heatmap
plt.figure(figsize=(12, 6))
shap.plots.heatmap(shap_values, max_display=20, show=False)
plt.title("SHAP Heatmap - Global", pad=12)
plt.tight_layout()
plt.show()

# -------------------------
# 4) LIME (local explanation)
# -------------------------
from lime.lime_tabular import LimeTabularExplainer

feature_names = list(X.columns)

# categorical indices heuristic
cat_features = [
    i for i, c in enumerate(X.columns)
    if (str(X[c].dtype) == "object" or str(X[c].dtype).startswith("category") or str(X[c].dtype) == "bool")
]

explainer_lime = LimeTabularExplainer(
    training_data=np.array(X_bg),
    feature_names=feature_names,
    class_names=["class_0", "class_1"] if predict_proba_fn(X.iloc[:2]).shape[1] == 2 else None,
    categorical_features=cat_features if len(cat_features) else None,
    discretize_continuous=True,
    mode="classification"
)

exp = explainer_lime.explain_instance(
    data_row=np.array(x0.iloc[0]),
    predict_fn=predict_proba_fn,
    num_features=12
)

# Plot 11: LIME local plot
fig = exp.as_pyplot_figure()
fig.set_size_inches(10, 6)
plt.title("LIME Local Explanation - first row", pad=12)
plt.tight_layout()
plt.show()

# Plot 12: LIME text output
print("\nLIME explanation (feature -> weight):")
for k, v in exp.as_list():
    print(f"{k:>45s} : {v:+.4f}")

# -------------------------
# 5) Diagnostics + model-agnostic XAI
# -------------------------
if y is not None:
    y_num = y.copy()

    # map object labels to 0/1 if binary
    if y_num.dtype == "object":
        uniq = sorted(y_num.dropna().unique().tolist())
        if len(uniq) == 2:
            mapping = {uniq[0]: 0, uniq[1]: 1}
            y_num = y_num.map(mapping)

    y_num = pd.to_numeric(y_num, errors="coerce")
    valid = y_num.notna()

    proba = predict_proba_fn(X.loc[valid])[:, pos_idx]
    pred = (proba >= 0.5).astype(int)

    # Plot 13: Confusion matrix
    plt.figure(figsize=(7, 6))
    ConfusionMatrixDisplay.from_predictions(y_num.loc[valid], pred, values_format="d")
    plt.title("Confusion Matrix (threshold=0.5)", pad=12)
    plt.tight_layout()
    plt.show()

    # Plot 14: ROC curve
    plt.figure(figsize=(7.5, 6))
    RocCurveDisplay.from_predictions(y_num.loc[valid], proba)
    plt.title("ROC Curve", pad=12)
    plt.tight_layout()
    plt.show()

    # Plot 15: Precision-Recall curve
    plt.figure(figsize=(7.5, 6))
    PrecisionRecallDisplay.from_predictions(y_num.loc[valid], proba)
    plt.title("Precision-Recall Curve", pad=12)
    plt.tight_layout()
    plt.show()

    # Plot 16: Calibration curve
    frac_pos, mean_pred = calibration_curve(y_num.loc[valid], proba, n_bins=10, strategy="quantile")
    plt.figure(figsize=(7.5, 6))
    plt.plot(mean_pred, frac_pos, marker="o")
    plt.plot([0, 1], [0, 1], linestyle="--")
    plt.xlabel("Mean predicted probability")
    plt.ylabel("Fraction of positives")
    plt.title("Calibration Curve", pad=12)
    plt.tight_layout()
    plt.show()

# Plot 17: Permutation importance
if y is not None:
    uniq = pd.Series(y_num.dropna().unique())
    scoring = "roc_auc" if len(uniq) == 2 else "accuracy"
    pi = permutation_importance(
        model, X.loc[valid], y_num.loc[valid],
        n_repeats=10, random_state=42, scoring=scoring, n_jobs=-1
    )
else:
    baseline = predict_proba_fn(X)[:, pos_idx].mean()

    def scoring_callable(estimator, Xp, y_dummy):
        p = predict_proba_fn(Xp)[:, pos_idx]
        return -np.mean((p - baseline) ** 2)

    pi = permutation_importance(
        model, X, np.zeros(len(X)),
        n_repeats=8, random_state=42, scoring=scoring_callable, n_jobs=-1
    )

imp = pd.Series(pi.importances_mean, index=X.columns).sort_values(ascending=False).head(20)
plt.figure(figsize=(11, 6))
imp.iloc[::-1].plot(kind="barh")
plt.title("Permutation Importance (Top 20) - Global", pad=12)
plt.xlabel("Mean importance")
plt.tight_layout()
plt.show()

# Plot 18-19: PDP + ICE for top 2 features
top2 = list(imp.index[:2])
for feat in top2:
    try:
        fig, ax = plt.subplots(figsize=(9.5, 6))
        PartialDependenceDisplay.from_estimator(model, X, [feat], kind="average", grid_resolution=30, ax=ax)
        ax.set_title(f"Partial Dependence (PDP) - {feat}", pad=12)
        plt.tight_layout()
        plt.show()

        fig, ax = plt.subplots(figsize=(9.5, 6))
        PartialDependenceDisplay.from_estimator(
            model, X, [feat],
            kind="individual", subsample=min(120, len(X)), grid_resolution=30, ax=ax
        )
        ax.set_title(f"ICE (Individual Conditional Expectation) - {feat}", pad=12)
        plt.tight_layout()
        plt.show()
    except Exception as e:
        print(f"Skipped PDP/ICE for '{feat}' due to: {e}")

print("\nDone. Generated SHAP, LIME, and additional explainability/diagnostic plots.")
```
